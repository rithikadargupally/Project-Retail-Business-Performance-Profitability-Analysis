```python
#1profit by category bar chart

category_profit = df.groupby("Category")["Profit"].sum()

category_profit.plot(kind="bar", figsize=(8,5))

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
# Monthly Sales Trend(line chart)

df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(
    df["Order Date"].dt.strftime("%b")
)["Sales"].sum()

# arrange months correctly
month_order = [
    "Jan","Feb","Mar","Apr",
    "May","Jun","Jul","Aug",
    "Sep","Oct","Nov","Dec"
]

monthly_sales = monthly_sales.reindex(month_order)

monthly_sales.plot(
    kind="line",
    figsize=(10,5),
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

monthly_sales
```

```python
#2profit by sub category
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values()

subcategory_profit.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.show()
```

```python
#3 top 10 product by cells
top_products_sales = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_sales.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Product Name")
plt.ylabel("Sales")

plt.xticks(rotation=75)

plt.show()
```

```python
#loss making product charts
loss_products = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
    .head(10)
    .sort_values(ascending=True)
)

loss_products.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Top 10 Loss-Making Products")
plt.xlabel("Product Name")
plt.ylabel("Profit Loss")

plt.xticks(rotation=75)

plt.show()
```
